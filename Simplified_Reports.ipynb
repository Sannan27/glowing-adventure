```python
# Import

from google.colab import files
import pandas as pd
```

```python
nwd_raw_file = pd.read_excel('/content/NWD Raw File 2.xlsx')
```

```python
nwd_raw_file.head()
```

```python
nwd_raw_file_revised = nwd_raw_file.copy()
nwd_raw_file_revised = nwd_raw_file_revised.groupby(['Day','Product'], as_index=False)[['Subscription Subs - Charged', 'Subscription Count - Charged','Subscription Revenue']].sum()
nwd_raw_file_revised.to_excel('offers_pd.xlsx')

#remaining columns 'Minutes - Product Usage - OnNet', 'Minutes - Product Usage - OffNet', 'Volume Product Usage','Minutes - Product Usage - PTCL', 'SMS Count - Product Usage'

```
