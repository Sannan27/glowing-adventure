```python
# Import

from google.colab import files
import pandas as pd
```

```python
nwd_raw_file = pd.read_excel('/content/Raw Data - 18 till 19.xlsx')
```

```python
nwd_raw_file.head()
```

```python
nwd_raw_file_revised = nwd_raw_file.copy()
nwd_raw_file_revised = nwd_raw_file_revised.groupby(['Day','Product'], as_index=False)[['Subscription Count - Charged','Subscription Subs - Charged','Subscription Revenue','Volume Product Usage', 'Minutes - Product Usage - OnNet','Minutes - Product Usage - PTCL', 'Minutes - Product Usage - OffNet',  'SMS Count - Product Usage']].sum()
nwd_raw_file_revised.to_excel('offers_pd.xlsx')

#remaining columns 'Minutes - Product Usage - OnNet', 'Minutes - Product Usage - OffNet', 'Volume Product Usage','Minutes - Product Usage - PTCL', 'SMS Count - Product Usage'

```

```python
import pandas as pd

regional_raw_file = pd.read_excel('/content/Regional_View_Raw.xlsx',sheet_name='Raw Data')

regional_raw_file2 = regional_raw_file.copy()
regional_raw_file2.head()
#regional_raw_file2['DT'] = pd.to_datetime(regional_raw_file2['DT'])
regional_raw_file2 = regional_raw_file2.groupby(['DT'],as_index=False)[['Daily_Minutes_OG_Offnet','Daily_Minutes_OG_Onnet','Daily_Volume_Gbs','Daily_Total_Revn_Amt','Total_Recharge_Amt']].sum()
regional_raw_file2.to_excel('regional_pd.xlsx')
```
